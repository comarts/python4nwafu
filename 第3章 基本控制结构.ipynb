# 第3章 基本控制结构

🏍️本章讲解的顺序执行、条件分支、循环迭代和异常处理是构建 Python 程序核心流程的基石。下面是一个具有复杂控制流的示例函数：

```
def complex_data_processor(input_config: Any) -> list[int, str]:
    """一个具有复杂控制流的示例函数."""
    if not isinstance(input_config, dict):
        return []

    result_list = []

    try:
        for key, value in input_config.items():
            if key.startswith('_'):
                result_list.append(f'Found_hidden_{key}')
            elif key.endswith('meta'):
                result_list.append(f'Found_meta_{key}')
            if isinstance(value, int):
                if value > 0:
                    result_list.append(value * 2)
                else:
                    result_list.append(0)
            elif isinstance(value, str):
                length = len(value)
                if length < 3:
                    result_list.append(f'Short:{value}')
                elif length == 5:
                    result_list.append('Exact5')
                else:
                    result_list.append(f'Long_{length}')
            else:
                result_list.append(str(value))

            match value:
                case int():
                    if value % 2 == 0:
                        result_list.append(f'Even_{value}')
                    else:
                        result_list.append(f'Odd_{value}')
                case str():
                    match len(value):
                        case n if n > 10:
                            result_list.append('VeryLongString')
                        case n if n >= 5:
                            result_list.append('MediumString')
                        case _:
                            result_list.append('ShortString')
                case _:
                    float_val = float(value) if isinstance(value, str) else value
                    result_list.append(float_val)
        if len(result_list) > 10:
            return [x for x in result_list if isinstance(x, int)]
    #except KeyError:
    #    return ['Processing_Failed_Key']
    except Exception:
        return ['Generic_Error_Caught']

    return result_list
```

**🐌圈复杂度——一种评估程序逻辑复杂度的指标**

Ruff 使用`圈复杂度`评判函数复杂程度，超出内置的阈值 10，程序告警。

圈复杂度（Cyclomatic Complexity），又称 McCabe 度量，是软件工程中用于衡量程序源代码逻辑复杂度的一个量化指标。圈复杂度表示的是程序中**线性无关路径的数量**。换句话说，它代表了为了测试程序中的每一条独立执行路径，最少需要设计多少个测试用例。在控制流图（Control Flow Graph）中，它也对应于图中的区域数量或闭环数量。

看起来有点复杂，我们使用基于决策点（最简单常用）方法来计算：
$$M = P + 1$$

*   P：程序中独立判断语句的数量（如 `if`、`while`、`for`、`case`、`except`）。
*   含义：每多一个判断，就多一条可能的路径。

complex_data_processor() 函数共有 16 个 `for`、`if` 和 `try`，所以它的圈复杂度 $M = P + 1 = 17$。

复杂的程序控制结构会导致诸多问题，摘选如下：
1. 代码可读性降低与认知负荷增加，难以快速判断程序在特定输入下的行为路径；
2. 缺陷密度与 Bug 隐患呈上升趋势，且修复成本随着版本迭代急剧增加；
3. 重构困难，当需求变更时，由于控制流错综复杂，牵一发而动全身，导致系统僵化；
4. 异常处理机制如果嵌入在复杂的控制流中，极易被绕过或误触发；

🏍️⚠️复杂的分支结构与循环，极易将我们拖入**代码维护**的泥潭；📌与其事后补救，不如事先规划，通过采用 Pythonic 惯用写法（如列表推导、高阶函数）、复用成熟的第三方工具、并将逻辑拆分为单一职责的独立模块，从而让程序回归简洁优雅的 `Zen of Python`。

#### 🔍Why Python 3.10+ Needs `match`?

多分支结构可以使用 match case 语句实现。

🔖以下内容源自 \[Python 文档——Python 的新变化\] » Python 3.10 有什么新变化 » PEP 634：结构化模式匹配

Python 3.10 版本增加了采用模式加上相应动作的 `match` 语句和 `case` 语句的形式的结构化模式匹配。 模式由序列、映射、基本数据类型以及类实例构成。 模式匹配使得程序能够从复杂的数据类型中提取信息、根据数据结构实现分支，并基于不同的数据形式应用特定的动作。

##### 🔜语法与操作

模式匹配的通用语法如下:
```
match subject:
    case <pattern_1>:
        <action_1>
    case <pattern_2>:
        <action_2>
    case <pattern_3>:
        <action_3>
    case _:
        <action_wildcard>
```
match 语句接受一个表达式并将其值与以一个或多个 case 语句块形式给出的一系列模式进行比较。 具体来说，模式匹配的操作如下：

1. 使用具有特定类型和形状的数据 (subject)
1. 针对 subject 在 match 语句中求值
1. 从上到下对 subject 与 case 语句中的每个模式进行比较直到确认匹配到一个模式。
1. 执行与被确认匹配的模式相关联的动作。
1. 如果没有确认到一个完全的匹配，则如果提供了使用通配符 _ 的最后一个 case 语句，则它将被用作已匹配模式。 如果没有确认到一个完全的匹配并且不存在使用通配符的 case 语句，则整个 match 代码块不执行任何操作。。

##### 📢声明性方式

读者可能是通过 C, Java 或 JavaScript (以及其他许多语言) 中的 switch 语句将一个目标 (数据对象) 与一个字面值 (模式) 进行匹配的简单例子了解到模式匹配的概念的。 switch 语句常常被用来将一个对象/表达式与包含在 case 语句中的字面值进行比较。

更强大的模式匹配例子可以在 Scala 和 Elixir 等语言中找到。 这种结构化模式匹配方式是“声明性”的并且会显式地为所要匹配的数据指定条件（模式）。

虽然使用嵌套的“if”语句的“命令性”系列指令可以被用来完成类似结构化模式匹配的效果，但它没有“声明性”方式那样清晰。 相反地，“声明性”方式指定了一个匹配所要满足的条件，并且通过其显式的模式使之更为易读。 虽然结构化模式匹配可以采取将一个变量与一个 case 语句中的字面值进行比较的最简单形式来使用，但它对于 Python 的真正价值在于其针对目标类型和形状的处理操作。

##### 🔟简单模式：匹配一个字面值

让我们把这个例子看作是模式匹配的最简单形式：一个值，即主词，被匹配到几个字面值，即模式。在下面的例子中，status 是匹配语句的主词。模式是每个 case 语句，字面值代表请求状态代码。匹配后，将执行与该 case 相关的动作:

````python
def http_error(status):
    match status:
        case 400:
            return "Bad request"
        case 404:
            return "Not found"
        case 418:
            return "I'm a teapot"
        case _:
            return "Something's wrong with the internet"

print(http_error(418))   
print(http_error(500))
````

如果传给上述函数的 status 为 418，则会返回 "I'm a teapot"。 如果传给上述函数的 status 为 500，则带有 `_` 的 case 语句将作为通配符匹配，并会返回 "Something's wrong with the internet"。 请注意最后一个代码块：变量名 `_` 将作为 通配符 并确保目标将总是被匹配。 `_` 的使用是可选的。

你可以使用 | （“ or ”）在一个模式中组合几个字面值:
```
case 401 | 403 | 404:
    return "Not allowed"
```

````python
def http_error(status):
    match status:
        case 401 | 403 | 404:
            return "Not allowed"
        case _:
            return "Something's wrong with the internet"

print(http_error(401))   
print(http_error(500))
````

###### ✳️无通配符的行为

如果我们修改上面的例子，去掉最后一个 case 块，这个例子就变成:

````python
def http_error(status):
    match status:
        case 400:
            return "Bad request"
        case 404:
            return "Not found"
        case 418:
            return "I'm a teapot"

print(http_error(500))          
````

如果不在 case 语句中使用 `_`，可能会出现不存在匹配的情况。如果不存在匹配，则行为是一个 no-op。例如，如果传入了值为 500 的 status ，就会发生 no-op。

##### ❓带有字面值和变量的模式

模式可以看起来像解包形式，而且模式可以用来绑定变量。在这个例子中，一个数据点可以被解包为它的 x 坐标和 y 坐标:

````python
# point 是一个 (x, y) 元组
point = (1, 2)
match point:
    case (0, 0):
        print("Origin")
    case (0, y):
        print(f"Y={y}")
    case (x, 0):
        print(f"X={x}")
    case (x, y):
        print(f"X={x}, Y={y}")
    case _:
        raise ValueError("Not a point")
````

第一个模式有两个字面值 (0, 0)，可以看作是上面所示字面值模式的扩展。 接下来的两个模式结合了一个字面值和一个变量，而变量 绑定 了一个来自主词的值 (point)。 第四种模式捕获了两个值，这使得它在概念上类似于解包赋值 (x, y) = point。

##### 模式和类

如果你使用类来结构化你的数据，你可以使用类名后面跟一个与构造器对应的参数列表的模式。 这种模式可以将实例属性捕获到变量中:

````python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

def location(point):
    match point:
        case Point(x=0, y=0):
            print("Origin is the point's location.")
        case Point(x=0, y=y):
            print(f"Y={y} and the point is on the y-axis.")
        case Point(x=x, y=0):
            print(f"X={x} and the point is on the x-axis.")
        case Point():
            print("The point is located somewhere else on the plane.")
        case _:
            print("Not a point")

location(Point(1, 2))
location((1, 2))
````

###### 📍带有位置参数的模式
你可以在某些为其属性提供了排序的内置类（例如 dataclass）中使用位置参数。 你也可以通过在你的类中设置 `__match_args__` 特殊属性来为模式中的属性定义一个专门的位置。 如果它被设为 ("x", "y")，则以下模式均为等价的（并且都是将 y 属性绑定到 var 变量）:
```
Point(1, var)
Point(1, y=var)
Point(x=1, y=var)
Point(y=var, x=1)
```
##### 🗂️嵌套模式
模式可以任意地嵌套。 例如，如果我们的数据是由点组成的短列表，则它可以这样被匹配:

````python
points = [Point(0, 3), Point(0, 2)]

match points:
    case []:
        print("No points in the list.")
    case [Point(0, 0)]:
        print("The origin is the only point in the list.")
    case [Point(x, y)]:
        print(f"A single point {x}, {y} is in the list.")
    case [Point(x=0, y=y1), Point(x=0, y=y2)]:
        print(f"Two points on the Y axis at {y1}, {y2} are in the list.")
    case _:
        print("Something else is found in the list.")
````

##### 🤯复杂模式和通配符
到目前为止，这些例子仅在最后一个 `case` 语句中使用了 `_`。 但通配符可以被用在更复杂的模式中，例如 `('error', code, _)`。 举例来说:
```
match test_variable:
    case ('warning', code, 40):
        print("A warning has been received.")
    case ('error', code, _):
        print(f"An error {code} occurred.")
```        
在上述情况下，`test_variable` 将可匹配 `('error', code, 100)` 和 `('error', code, 800)`。

##### 🤔约束项
我们可以向一个模式添加 `if` 子句，称为“约束项”。 如果约束项为假值，则 match 将继续尝试下一个 case 语句块。 请注意值的捕获发生在约束项被求值之前:

````python
point = Point(1, 1)

match point:
    case Point(x=x, y=y) if x == y:
        print(f"The point is located on the diagonal Y=X at {x}.")
    case Point(x=x, y=y):
        print(f"Point is not on the diagonal.")
````

##### 🤷‍其他关键特性
一些其他关键特性:

* 类似于解包赋值，元组和列表模式具有完全相同的含义，而且实际上能匹配任意序列。 从技术上说，目标必须为一个序列。 因而，一个重要的例外是模式不能匹配迭代器。 而且，为了避免一个常见的错误，序列模式不能匹配字符串。

* 序列模式支持通配符: `[x, y, *rest]` 和 `(x, y, *rest)` 的作用类似于解包赋值中的通配符。 在 `*` 之后的名称也可以为 `_`，因此 `(x, y, *_)` 可以匹配包含两个条目的序列而不必绑定其余的条目。

* 映射模式: `{"bandwidth": b, "latency": l}` 会从一个字典中捕获 "bandwidth" 和 "latency" 值。 与序列模式不同，额外的键会被忽略。 也支持通配符 `**rest`。 （但 `**_` 是冗余的，因而不被允许。）

* 子模式可使用 `as` 关键字来捕获:
```
case (Point(x1, y1), Point(x2, y2) as p2): ...
```
    x1, y1, x2, y2 等绑定就如你在没有 as 子句的情况下所期望的，而 p2 会绑定目标的整个第二项。

* 大多数字面值是按相等性比较的。 但是，单例对象 True, False 和 None 则是按标识号比较的。

* 命名常量也可以在模式中使用。 这些命名常量必须为带点号的名称以防止常量被解读为捕获变量:

````python
from enum import Enum
class Color(Enum):
    RED = 0
    GREEN = 1
    BLUE = 2

color = Color.GREEN
match color:
    case Color.RED:
        print("I see red!")
    case Color.GREEN:
        print("Grass is green")
    case Color.BLUE:
        print("I'm feeling the blues :(")
````

## 循环结构

### 循环结构解决的问题类型

改写课本示例，创建一个存放学生成绩的文件——data/scores.csv：
```CSV
gs, yy, cx
85, 92, 78
76, 88, 90
91, 85, 82
70, 65, 79

```
第一行为表头，后续行为每位同学的成绩，最后刻意增加一个空白行，模拟异常数据。

使用循环结构读出每行数据，切分为三门课成绩，转换成浮点型，求出平均成绩并打印：

````python
for line in open('data/scores.csv', 'r', encoding='utf-8'):
    parts = line.strip().split(',')
    if len(parts) < 4 or '"ID"' == parts[0]:
        continue
    gs, yy, cx = float(parts[1]), float(parts[2]), float(parts[3])
    print((gs + yy + cx) / 3)
````

## 异常处理

### 异常的概念及产生原因

Python 异常表示程序运行中遇到的意外中断或特殊条件。根据设计需求，它既可以由函数内部捕获修复，也可以向上传递至调用者进行决策和处理。

### 异常处理机制

❗以下内容使用 Qwen3.5:9B 生成

Python 的异常机制深刻诠释了软件设计中的**契约精神**。它定义了函数在预期失败条件下的行为边界，明确了向调用者传递错误状态的义务。因此，异常不仅是一种控制流跳转手段，更是模块间界定“可用与不可用”边界的关键协议。

以**餐厅点餐用餐**为例：厨师（开发者）根据食材和厨具能力烹饪菜品（API 设计），顾客（调用者）需要依据菜单（文档）知晓潜在风险。如果存在食材过敏、后厨设备故障或无法制作的特殊情况，厨房必须明确告知食客（抛出异常）。如果后厨隐瞒隐患导致食客健康受损，这违背了餐饮服务的信任原则；同理，Python 函数也应在特定条件下触发异常以中断流程，避免污染全局状态。

1. 创建函数：编写“菜单与过敏原提示”（定义契约边界）
创建函数的目的不仅是提供服务，更是确立服务的风险告知范围。如同餐厅会在菜单显眼处标注“本店不提供辣菜”或“含坚果预警”，开发者也应在 `Docstring` 的 `Raises` 章节中，像厨师长一样严谨地声明可能抛出的异常类型及触发条件：
*   明确预期失败： 无效参数对应 `ValueError`；设备故障（系统错误）对应 `RuntimeError`。
*   文档即契约： 隐瞒潜在风险等同于“以次充好”，违背了 API 设计的诚实原则。调用者有权依据文档拒绝食用（不调用），而不应在不知情中遭遇“食物中毒”（未预期的崩溃）。

2. 调用函数：履行“用餐保障义务”（执行风险控制）
函数调用者是契约的另一方责任主体，在调用函数时，有义务通过 `try...except` 结构去识别并处理违约行为。
*   善后与兜底： 捕获异常后进行重试、记录日志或降级服务，确保系统“能继续吃下一顿”；
*   违约后果： 若未对异常进行妥善处理（即忽略报错），程序将因无法处理的错误而终止运行。这违背了“契约”中保障服务稳定性的初衷，等同于食客在餐厅中毒后选择直接离场甚至引发法律纠纷，破坏了整个用餐体验的完整性。

### ✍️伴读札记

#### 🔥EAFP Or LBYL

❗以下内容使用 Qwen3.5:9B 生成

**EAFP**（Easier to Ask Forgiveness than Permission）与 **LBYL**（Look Before You Leap）是两种不同的编程风格，主要用于处理代码中的错误预期和数据访问。这两种风格在 Python 社区中争论尤为激烈，但也适用于其他语言。

简单来说：
*   **EAFP**: 先做，错了再处理 (Try-Except)
*   **LBYL**: 先看条件，再做 (If Check-Then Do)

以下是详细的对比分析：

1. 核心定义与代码示例

**EAFP (先做后错)**

哲学：假设操作会成功。如果遇到错误，捕获异常并处理。

这是 Python 社区推崇的“默认”风格（Python Zen）。它倾向于减少嵌套逻辑，使代码流程更线性。

```python
# EAFP 示例：获取字典中的键
try:
    value = my_dict['key']  # 直接尝试访问
except KeyError:
    value = None            # 处理错误情况
print(value)
```

**LBYL (先看后跳)**

哲学：在操作前检查条件。如果条件满足才执行，否则跳过或返回默认值。

常见于 Java、C# 等强类型语言，或者 Python 中对特定数据结构访问的优化写法。

```python
# LBYL 示例：获取字典中的键
if 'key' in my_dict:      # 先检查是否存在
    value = my_dict['key'] # 再执行操作
else:
    value = None          # 处理不存在的情况
print(value)
```
2. 优缺点详解

**EAFP (先做后错)**

*   ✅ 优点：
    *   更 Pythonic：符合“优雅优于明确”的哲学，代码通常更简洁。
    *   避免嵌套：不需要层层 `if` 判断，业务逻辑一目了然（Happy Path）。
    *   利用异常机制：Python 的异常处理经过高度优化，对于罕见错误效率很高。
*   ❌ 缺点：
    *   调试困难：如果捕获过于宽泛的异常，可能导致“吞掉”真正的错误。
    *   性能开销：如果频繁报错（例如循环中访问不存在的数据），抛出和捕获异常比简单的布尔判断要慢得多。

**BYL (先看后跳)**
*   ✅ 优点：
    *   显式意图：明确告诉读者“这一步需要检查”，避免意外。
    *   性能优化潜力：在高性能场景下，如果错误率较高，直接判断比处理异常更快。
    *   安全性：在某些并发或锁场景中，先检查再操作可能更安全（尽管 Python 不推荐显式锁）。

*   ❌ 缺点：
    *   “钻入地狱” (The YAGNI Problem)：当条件复杂时，`if-else` 会写得非常冗长。
    *   双重计算：例如 `if key in dict: val = dict[key]`，在 CPython 中需要两次哈希查找（一次检查存在性，一次取值），而直接取值的开销只算一次（虽然失败了）。

3. 性能考量 (关键陷阱)

这是一个经常被误解的点：**“异常处理很慢”是真的吗？**

*   罕见错误 (Rare Errors):
    *   如果绝大多数情况都成功，EAFP 通常更快。因为 Python 中的 `try` 块在成功时开销极小。而 LBYL 需要显式检查条件。
*   频繁错误 (Frequent Errors):
    *   如果操作经常失败（例如遍历一个可能包含 `None` 的列表），LBYL 更快。因为抛出和捕获异常对象涉及内存分配，成本远高于简单的比较运算。

4. Python 社区的最佳实践

在 Python 中，**EAFP 通常是首选**。

Python 官方文档 (Effective Python) 和 Guido van Rossum 多次提到：**"Try to do the thing first, handle exceptions if it fails."**

为什么？因为可读性。想象以下代码：

*   LBYL (糟糕的嵌套):
    ```python
    if obj is not None:
        if hasattr(obj, 'data'):
            if obj.data is not None:
                print(obj.data)
            else:
                pass # 处理空数据
        else:
            pass # 没属性
    else:
        pass # 对象本身是 None
    ```

*   EAFP (清晰的线性):
    ```python
    try:
        value = obj.data
    except AttributeError:
        value = "No data"
    print(value)
    ```

**一句话总结：**
在 Python 中，除非你非常清楚异常处理会带来严重的性能瓶颈或代码逻辑极其复杂，否则请坚持使用 **EAFP**。

# 第6章 自定义函数
🏍️本章从模块化编程角度讲解如何定义和使用函数。

**告别代码复制，迈向模块化编程**

❗以下内容使用 Qwen3.5:9B 生成

要构建模块化的系统，自定义函数必须遵循特定的设计规范。以下是现代 Python 工程中的最佳实践范式：


1. 单一职责原则 (Single Responsibility Principle)

一个函数只做一件事，且做好这件事。如果一个函数既负责计算又负责打印结果，它就是一个“黑盒”，难以测试和维护。

*   Bad: `process_data()`（内部包含读取、清洗、转换、保存五个步骤）
*   Good:
    *   `read_file(path)` -> 返回流对象
    *   `clean_data(rows)` -> 返回列表
    *   `save_to_db(data)` -> None

2. 显式优于隐式 (Explicit is Better than Implicit)

Python 是动态类型语言，但在定义函数时，应当明确表达意图。

*   参数命名：使用蛇形命名法（snake_case），避免缩写或模糊词汇。
    *   `calc_total_price(items)` vs `cal_sum(i)`
*   默认参数陷阱：尽量避免在默认参数中绑定可变对象（如列表、字典）。
    *   ❌ `def add_to_list(item, mylist=[]):` (每次调用都修改同一个 list 引用)
    *   ✅ `def add_to_list(item, mylist=None): if mylist is None: mylist = []`

3. 类型提示与契约 (Type Hints & Contract)

在 Python 3.5+ 中，使用 Type Hints 是现代模块化编程的标配。它不仅是给 IDE 提供智能补全的工具，更是函数接口的“契约文档”。

```python
from typing import List, Optional

# 输入：一个字符串列表和一个整数；输出：计算后的总和；失败抛异常
def calculate_average(numbers: List[float], count: int) -> float:
    if len(numbers) != count:
        raise ValueError("Input length mismatch")
    return sum(numbers) / count

# 处理可能返回 None 的情况，调用方需知晓这一点
get_user_id() -> Optional[int]: 
```

4. 无副作用函数 (Side-Effect Free Functions)

在单元测试（Testability）中，纯函数是黄金标准。如果一个函数的返回值依赖于外部状态（如读取全局变量、修改文件），它很难被隔离测试。

*   理想范式： 输入 -> 处理 -> 输出（不修改传入的参数）。
*   反例： `def modify_data(data): data.append(...)`
*   修正： `def append_to_data(list_ref, item): return list_ref + [item]` (返回新对象)

5. 文档字符串 (Docstrings)

函数不仅仅是代码，它是接口。每个公共函数必须包含符合规范的 Docstring（通常采用 Google 风格或 NumPy 风格）。

```python
def validate_email(email: str, domain_whitelist: List[str]) -> bool:
    """
    验证用户邮箱地址的有效性。

    Args:
        email (str): 待验证的邮箱字符串，不能为空。
        domain_whitelist (List[str]): 允许接收邮件的后缀列表（如 ['gmail.com']）。

    Returns:
        bool: True 如果有效且域名在白名单内；False 否则。

    Raises:
        ValueError: 当输入包含非法字符时抛出此异常。
    """
```

6. 错误处理与边界值 (Error Handling)

不要使用 `pass` 或空语句吞掉异常。模块化函数必须明确告知调用者何时会失败。

*   原则： 优先使用 `raise Exception(...)` 来表明业务逻辑错误，而不是返回错误码（如 -1）。

*   输入验证：在函数入口处进行参数校验（Guard Clauses）。避免深层嵌套的 if-else。

```python
def safe_divide(a: float, b: float) -> float:
    # Guard Clause (卫语句)
    if a is None or b is None:
        raise TypeError("Arguments cannot be None")
    if b == 0:
        raise ZeroDivisionError("Denominator cannot be zero")
    
    return a / b
```

✅ 为什么不建议一次性完成？

#### 👥该不该用可变参数

❗以下内容使用 Qwen3.5:9B 生成

在 Python 中是否“应该”使用可变参数（`*args` 和 `**kwargs`），取决于具体的应用场景和设计意图。它们本身是合法且常用的语法工具，但滥用会导致代码可读性差、维护成本高甚至出现难以调试的 bug。以下从多个维度为你分析其适用性与最佳实践：

✅ 什么时候应该使用可变参数？

1. 函数接口不确定

适用于调用者可以传入数量或类型不确定的参数，例如日志记录器、通用处理器等：

```python
def log(message, *args):
    print(f"{message}: {', '.join(str(a) for a in args)}")

log("Error", "file not found", "code: 500")
```

2. 函数式封装/组合工具（如装饰器）

在函数式编程中，可变参数可用于灵活构建或传递多个参数：

```python
def apply(func, *args):
    return func(*args)

print(apply(sum, [1, 2, 3]))
```

3. 兼容旧接口/第三方库

当需要与外部 API、遗留代码或不确定签名的模块对接时，可变参数有助于灵活性：

```python
def call_api(*args):
    # 转发参数给外部函数
    return target_function(*args)
```

❌ 什么时候应避免使用？

1. 接口语义不清晰

如果函数的参数含义不明确或数量不确定，会严重降低可读性：

```python
# ⚠️ 不推荐：调用者无法了解参数含义
def process_data(*args):
    ...

process_data(1, "a", True)  # ❓这代表什么？
```

2. 可维护性与调试困难

可变参数使参数来源模糊，难以追踪错误信息、单元测试或文档说明。


🛠️ 实用建议总结

- ✅ 优先使用明确的函数签名（命名参数、默认值）。
- ⚠️ 仅在接口不固定或需高度灵活性时使用 `*args` / `**kwargs`。
- 🔁 配合类型注解（如 `typing.Union`, `Optional`）增强可读性。
- 🧪 编写文档和测试用例，说明参数用途。
- 📝 对复杂函数添加详细文档注释。

#### 👑装饰器

❗以下内容使用 Qwen3.5:9B 生成

装饰器（Decorator）是一种用于为函数或方法添加额外功能，而不修改其原始代码的技术手段。它可以被理解为一种高级的函数包装机制，常用于增强函数的功能（如日志记录、权限控制、性能优化等），且无需更改原有函数逻辑。

基本结构

```python
def decorator_name(func):
    def wrapper(*args, **kwargs):
        # 在函数执行前后添加的功能
        result = func(*args, **kwargs)
        return result
    return wrapper

@decorator_name
def original_func():
    pass
```

装饰器原理图解
```python
# 等价于：
original_func = decorator_name(original_func)
```

下面代码定义了一个日志装饰器 `log_decorator`，其作用是在执行被装饰的函数前，打印出该函数的名称。

具体逻辑如下：
1.  装饰过程：通过 `@log_decorator` 将 `greet` 和 `my_add` 两个函数包装了。
2.  运行效果：调用这两个函数时，控制台会先输出日志信息（如“调用函数：greet"），随后执行原函
数的功能逻辑。

````python
def log_decorator(func):
    def wrapper(*args, **kwargs):
        print(f"调用函数: {func.__name__}")
        return func(*args, **kwargs)
    return wrapper

@log_decorator
def greet(name):
    print(f"你好, {name}!")

@log_decorator
def my_add(x, y,):
    return x + y

greet("Alice") 
my_add(1, 2)
````

在 Python 中，调用栈（Call Stack）是解释器内部用来跟踪函数执行状态的关键数据结构。它遵循后进先出 (LIFO, Last In First Out) 的原则。每一个进入的函数都会创建一个“栈帧”（Stack Frame），退出时销毁该帧。

下面是调用栈的内存结构视图 (Stack Structure)，这张图展示了当函数 A 调用 B，B 又调用 C 时，它们在内存中的堆叠状态。每个矩形代表一个 栈帧（Stack Frame）：

```mermaid
flowchart TD
    subgraph Stack [Memory Stack Area]
        direction TB
        
        %% 顶部是最新创建的帧 (Top of Stack)
        FrameC[<b>Frame C</b><br/>func_c<br/>local vars, args]:::active
        FrameB[<b>Frame B</b><br/>func_b<br/>local vars, args]:::active
        FrameA[<b>Frame A</b><br/>func_a<br/>local vars, args]:::active
        Main[<b>Main Frame</b><br/>__main__<br/>entry point]:::base
        
    end

    %% 连接关系表示调用链，但不占用空间，仅示意
    FrameC --- FrameB
    FrameB --- FrameA
    FrameA --- Main
    
    %% 样式定义
    classDef active fill:#ffcccc,stroke:#333,stroke-width:2px,color:black;
    classDef base fill:#e0f7fa,stroke:#333,stroke-width:1px,color:black;

    %%title 调用栈内存堆叠示意图 (LIFO)
```

图解说明：
*   底部 (`Main`)：程序入口，最先创建。
*   顶部 (`Frame C`)：最后被调用的函数，占据当前栈顶位置。
*   层级关系：只有当 `C` 执行完毕（返回），`Frame C` 才会被移除（Pop），控制权回到 `B`，以此类推。

函数调用的层级不是无限的，使用 `sys.getrecursionlimit()` 获取最大限制。`len(inspect.stack())` 用来获取当前调用栈的深度。所以，递归函数支持的最大递归层数是：`sys.getrecursionlimit() - len(inspect.stack())`。

以 `count_char()` 为例，它寻找字符串 `s` 中出现字符 `c` 的次数，递归次数为 `len(s) + 1`：
```
def count_char(s, c):
    if not s:
        return 0
    return (1 if s[0] == c else 0) + count_char(s[1:], c)
```

传入的字符串最大长度 $\text{max}(length) = \text{sys.getrecursionlimit}() - \text{len(inspect.stack}()) - 1$，代码的执行结果验证了我们的推测：

````python
import inspect
import sys

def count_char(s, c):
    if not s:
        return 0
    return (1 if s[0] == c else 0) + count_char(s[1:], c)

try:
    my_word = ("hellokitty"*(int(sys.getrecursionlimit()/10) + 1))
    my_word = my_word[:(sys.getrecursionlimit() - len(inspect.stack()) - 1)]
    print(count_char(my_word, 'l')) 
    my_word = my_word + 'l'
    print(count_char(my_word, 'l')) 
except RecursionError as e:
    print(f'{e}, length is {len(my_word)}')
````

这段代码是一个用于演示 Python 中递归函数执行和调用栈限制的示例程序。它展示了一个递归函数 `count_char` 如何统计字符串中某个字符的出现次数，并尝试在接近递归深度极限的情况下运行该函数，以验证系统的行为表现。

    *   能力要求：`os` 模块操作目录，`pandas` 处理表格，`openpyxl` 写 Excel。
    *   结论：只需要简单的函数和循环即可。

2.  数值计算与仿真后处理 (Numerical Analysis)
    *   场景：求解微分方程、拟合实验数据、运行简单的物理模型（如热传导、结构分析）。
    *   能力要求：调用 `numpy`（矩阵运算）和 `scipy`。
    *   结论：核心逻辑是数学公式的翻译，不需要掌握复杂的递归或装饰器，函数封装足够清晰即可
。

3.  可视化 (Visualization)
    *   场景：绘制实验曲线、三维点云图、流场热力图。
    *   能力要求：调用 `matplotlib` 或 `plotly` / `pyvista`。
    *   结论：库已经封装好了复杂的绘图逻辑，你只需要配置参数（函数输入），不需要写渲染代码
。

4.  科研复现与论文 (Reproducibility)
    *   场景：运行别人的代码、处理数据、展示结果。
    *   能力要求：环境管理（Conda/venv）、Jupyter Notebook。
    *   结论：这属于“工具使用”范畴，不需要深厚的编程架构知识。

只要你能写出整洁的脚本，并且懂得把复杂问题拆解成函数，熟练使用主流科学计算库，就足以支撑绝大多数科研实验数据处理、简易自动化办公以及本科/硕士阶段的项目需求。

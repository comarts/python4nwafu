1. Beautiful is better than ugly.

    优雅好过丑陋
    ——代码应当整洁美观，拒绝晦涩、混乱、堆砌的写法

1. Explicit is better than implicit.

    显式好过隐式
    ——代码意图必须直白，避免隐式转换、魔法逻辑、无意义简写

1. Simple is better than complex.

    简单好过复合
    ——能用简单逻辑实现，绝不引入复杂架构

1. Complex is better than complicated.

    复合好过复杂
    ——业务无法简化时，保证结构清晰，不堆砌纠缠的混乱逻辑

1. Flat is better than nested.

    扁平好过嵌套
    ——减少多层if、循环嵌套，扁平化代码更容易调试阅读

1. Sparse is better than dense.

    稀疏好过密集
    ——合理空行分隔逻辑块，拒绝一行堆砌多条业务逻辑

1. Readability counts.

    可读性很重要
    ——代码首先写给人阅读，其次才是给机器执行

1. Special cases aren't special enough to break the rules.

    特例也并不特殊到足以破坏规则
    ——统一规范优先级最高，不要为少量特殊场景随意破坏编码标准

1. Although practicality beats purity.

    即使要为了实用性而牺牲纯粹性
    ——标准不是教条，真实业务的实用性优先于理想化规范

1.  In the face of ambiguity, refuse the temptation to guess.

    面对太多可能，不要尝试猜测

1.  There should be one-- and preferably only one --obvious way to do it.

    应该有一种（且最好只有一种）明显的方式来做到这一点

    ——能简单直白解释的实现，大概率是优质方案

1978年，柯林汉与丹尼斯·里奇（C语言之父）合著的 \[C程序设计语言\]（The C Programming Language） 出版。这本书被誉为“K&R C”，是C语言事实上的标准。书中将“hello, world”程序作为第一个示例，随着这本书在程序员中的广泛流传，“Hello, World!” 也正式成为编程世界的传统，被无数编程语言和教程沿用至今。

我们的第一个脚本——src/python4nwafu/hello_world.py，逐一诠释了语句、数据、变量、表达式、函数和注释：
```
#!/usr/bin/env python3
"""Hello, World!
"""
import sys


def main():
    """主函数.
    
    """
    # 定义变量
    message = "Hello, World!"
    name = "Python Developer"
    year = 2026

    # 使用变量输出
    print(message)
    print(f"作者：{name}")
    print(f"当前年份：{year}")
    print(f"Python解释器路径: {sys.executable}")


if __name__ == "__main__":
    main()
```

1. `#!/usr/bin/env python3`
    
    1. `#!`
    
        识别标记，操作系统看见文件第一开头是 `#!`，就知道这是 shebang。
    1. `/usr/bin/env`
    
        env 是系统工具：自动在系统环境变量 `PATH` 里搜索程序。
    1. `python3`
        
        需要寻找的程序名称。

    1. 执行逻辑：
        
        操作系统调用 `/usr/bin/env` → env 根据 PATH 找到系统里 `python3` 的真实路径 → 启动 python3 运行脚本。
2. `import sys`：导入系统模块，用来获取 Python 解释器路径。
3. `def main():` 定义函数，封装主逻辑，规范写法，避免全局变量泛滥。
4. `f"字符串{变量}"`：Python3.6+ f-string 格式化字符串，简洁直观。
5. `sys.executable`：获取当前正在执行的 Python 程序完整路径。
6. `if __name__ == "__main__":`
    
    只有直接运行本文件时才执行 `main()`；被其他文件 import 导入时不会自动运行，是 Python 标准最佳实践。

#### 📦一切皆为对象

如果同学们学习过 CS 101 课程，或者了解过 C 语言，对 Python 整数类型或者浮点类型占用的内存空间或许产生疑惑，像 10 这样很小的整数明明一个字节就可以存储，在 Python 中为什么占用了 28 个字节？原因是在 Python 的世界里，整数、浮点数、字符串等等都是对象，

🔖源自\[ Python 文档——Python 语言参考手册\] » 3. 数据模型 » 3.1. 对象、值与类型

`对象` 是 Python 中对数据的抽象。 Python 程序中的所有数据都是由对象或对象间关系来表示的。 甚至代码本身也是由对象来表示的。

每个对象都有相应的标识号、类型和值。 一个对象被创建后它的 标识号 就绝不会改变；你可以将其理解为该对象在内存中的地址。 is 运算符比较两个对象的标识号是否相同；id() 函数返回一个代表其标识号的整数。

在 CPython 中，id(x) 就是存放 x 的内存的地址。

对象的类型决定该对象所支持的操作 (例如 "对象是否有长度属性？") 并且定义了该类型的对象可能的取值。type() 函数能返回一个对象的类型 (类型本身也是对象)。与编号一样，一个对象的 类型 也是不可改变的。

有些对象的 值 可以改变。值可以改变的对象被称为 可变对象；值不可以改变的对象就被称为 不可变对象。(一个不可变容器对象如果包含对可变对象的引用，当后者的值改变时，前者的值也会改变；但是该容器仍属于不可变对象，因为它所包含的对象集是不会改变的。因此，不可变并不严格等同于值不能改变，实际含义要更微妙。) 一个对象的可变性是由其类型决定的；例如，数字、字符串和元组是不可变的，而字典和列表是可变的。

对象绝不会被显式地销毁；然而，当无法访问时它们可能会被作为垃圾回收。允许具体的实现推迟垃圾回收或完全省略此机制 --- 如何实现垃圾回收是实现的质量问题，只要可访问的对象不会被回收即可。

CPython 目前使用带有 (可选) 延迟检测循环链接垃圾的引用计数方案，会在对象不可访问时立即回收其中的大部分，但不保证回收包含循环引用的垃圾。请查看 gc 模块的文档了解如何控制循环垃圾的收集相关信息。其他实现会有不同的行为方式，CPython 现有方式也可能改变。不要依赖不可访问对象的立即终结机制 (所以你应当总是显式地关闭文件)。

🏍️从 Python 语言层面，一切皆为对象。如果跳出 Python API 和解释器可见范围，进入 CPython 的解释器内部（例如查看 C 源码或通过 sys._getframe 等底层探针），确实存在一些不构成 Python 对象的数据类型。当然这远远超出我们的学习范围，谨记 **In Python, everything is an object.**

在 Python 编程中，“魔法数”（Magic Numbers）指的是程序中直接出现的、没有明显意义的数字或字符串值（如 `123`、`0xFF`、“用户权限：ADMINISTRATOR”等），它们缺乏明确上下文、维护成本高，属于常见的“坏味道”之一。

示例
```python
# 代码难以理解
threshold = 595 * 128
score = result / 2.71828
max_retry = 3 × (1 + 0.4)
if status == "active":
    config["timeout"] = 3600

# 硬编码的字符串作为状态标识
user_role = "admin"
```

问题
- 数值或字符串含义不明确（维护成本高）
- 增加修改成本，容易引入 bug
- 降低代码的可读性与可维护性
- 不利于团队协作和长期维护

如何识别魔法数？

特征 | 示例 
-:|-:
出现在逻辑判断中（如 `if i > 0.25`） | ✅ 可优化 
作为循环范围、阈值 | ✅ 应提取配置 
字符串常量用于状态表示（如 `'ERROR'`） | ✅ 使用枚举/常量类

#### 📚变量在命名空间，对象在栈

namespace -- 命名空间<sup>\[Python 文档——术语对照表\]</sup>

命名空间是存放变量的场所。命名空间有局部、全局和内置的，还有对象中的嵌套命名空间（在方法之内）。命名空间通过防止命名冲突来支持模块化。例如，函数 builtins.open 与 os.open() 可通过各自的命名空间来区分。命名空间还通过明确哪个模块实现那个函数来帮助提高可读性和可维护性。例如，random.seed() 或 itertools.islice() 这种写法明确了这些函数是由 random 与 itertools 模块分别实现的。

❗以下内容使用 Qwen3.5:9B 生成

在 Python 中，**变量（variable）** 与 **对象（object）** 位于不同的“内存层次”，理解它们的区别是掌握动态语言的重要基础。下面从“所在位置”的角度详细说明二者的差异：

---

📍 1. 所在位置不同

对象（Object）：
- 存储在堆（Heap）中独立分配。
- 包含对象数据和属性信息（如 int、string、list、custom class instance）。
- 在 Python 内部使用“引用计数”和垃圾回收机制管理其生命周期。

```python
a = [1, 2]   # list 是一个对象，存储在堆中
print(id(a)) # 返回对象的内存地址
```

变量（Variable）：
- 是命名空间中的**标签或引用**。
- 本质是容器指向某个对象的指针。
- 存储在全局字典、本地作用域等框架中（可视为“栈”区域）。

```python
b = a    # b 引用同一个对象，变量名只是标签
print(id(b)) == id(a)     # True：b和a指向同一内存对象
```

🏗️ 2. 内存结构示意图

```
┌─────────────── 栈帧（局部/全局命名空间） ────────────────┐
│ b → ──────────────┐                                       │
│ a → ────────────┐     ├──── Heap（对象存储区） ────────► [列表对象]
│     (变量名)    ↓    │  ┌─────────────────────────────┬───└┘
│     └─> 引用 ──────►   │      变量 a 和 b 均指向同一堆内存位置
└─────────────────────────────────────────────────────────┘
```

🔍 3. 判断变量与对象的例子

判断是否是对象（通过值）：

```python
# 所有 Python 中的值都封装在对象中
x = 10
type(x) => <class 'int'>         # int 类型的对象
isinstance(x, object)          # True，因为一切皆对象
print(type(x).__module__)      # 表示对象类型模块

```

判断是否指向相同对象（通过引用）：

```python
obj = [123]
copy_id = obj                  # copy_id 和 obj 是同个对象的变量名
obj is copy_id                 # True，因为变量引用的同一堆内存位置
id(obj) == id(copy_id)         # True

```

🔄 4. 对象与变量的关系 - 引用传递

在 Python 中，**所有赋值（=）都创建引用**。这意味着：

- **绑定操作 `=`** 不会复制对象内容，只是让变量指向同一堆内存对象。
- 修改可变对象会影响其他引用它的变量（如 list、dict）。
- 对于不可变对象（如 int、tuple），重新赋值会生成新对象并切换引用目标。

```python
a = [1, 2]
b = a       # b 引用与 a 相同的列表对象；[1,2] 是对象的值
print(a is b)   # True，指向相同堆内存

a, b = 'x', None    # a 和新值绑定；'x'是新对象
```

##### 🤫`__*`
    
类的私有名称。类定义时，此类名称以一种混合形式重写，以避免基类及派生类的 "私有" 属性之间产生名称冲突。详见 标识符（名称）。

在 Python 中，以两个下划线开头（例如 `__variable_name`）的属性称为“私有字段”。当类定义时会自动进行名称改写（Name Mangling），即 `ClassName.__private_attr` 会被转换为 `_ClassName__private_attr`。这能防止外部直接访问，起到一定程度上的命名保护和封装作用。    

````python
class Person:
    def __init__(self, name):
        self.name = name           # 普通属性（public）
        self.__age = 25            # 私有属性（private）

    def get_age(self):              # 控制外部访问的公共方法
        return f"My age is {self._Person__age}"

p = Person("Alice")
print(p.name)                     # Alice
print(p.get_age())                # My age is 25

# ❌ 不推荐直接访问私有属性：
try:
    print(p.__age)                 # 错误（可能失败）
except AttributeError as e:
    print(f"无法通过 __age 访问，提示：{e}")

# ✅ 实际路径是 _Person__age：
print(p._Person__age)             # 25
````

#### 💎特殊参数

🔖源自\[Python 文档——Python 教程\] » 4. 更多控制流工具 » 4.9.3. 特殊参数

默认情况下，参数可以按位置或显式关键字传递给 Python 函数。为了让代码易读、高效，最好限制参数的传递方式，这样，开发者只需查看函数定义，即可确定参数项是仅按位置、按位置或关键字，还是仅按关键字传递。

函数定义如下：
```
def f(pos1, pos2, /, pos_or_kwd, *, kwd1, kwd2):
      -----------    ----------     ----------
        |             |                  |
        |        位置或关键字   |
        |                                - 仅限关键字
         -- 仅限位置
```         
/ 和 * 是可选的。这些符号表明形参如何把参数值传递给函数：位置、位置或关键字、关键字。关键字形参也叫作命名形参。

* 位置或关键字参数
函数定义中未使用 / 和 * 时，参数可以按位置或关键字传递给函数。

* 仅位置参数
此处再介绍一些细节，特定形参可以标记为 仅限位置。仅限位置 时，形参的顺序很重要，且这些形参不能用关键字传递。仅限位置形参应放在 / （正斜杠）前。/ 用于在逻辑上分割仅限位置形参与其它形参。如果函数定义中没有 /，则表示没有仅限位置形参。

    / 后可以是 位置或关键字 或 仅限关键字 形参。

* 仅限关键字参数
把形参标记为 仅限关键字，表明必须以关键字参数形式传递该形参，应在参数列表中第一个 仅限关键字 形参前添加 *。

#### 📦模块、包和导入

在 Python 中，标准库是内置于语言环境中的大量功能模块，涵盖了从数据处理到网络通信等各个方面的使用场景。使用它们前通常需要 `import` 相关模块或函数。

模块是 Python 代码的基本组织单元，由`导入系统`创建，它或是通过 `import` 语句，或是通过调用 `importlib.import_module()` 和内置的 `__import__()` 等函数来唤起。 模块对象具有通过`字典`对象实现的命名空间（就是被定义在模块中的函数的 `__globals__` 属性所引用的字典）<sup>\[Python 文档——Python 语言参考手册\] » 3.2.9. 模块</sup>。

这个说法太专业，下面的说法简单易懂。

Python 把各种定义存入一个文件，在脚本或解释器的交互式实例中使用。这个文件就是`模块` ；模块中的定义可以`导入`到其他模块或`主 模块`（在顶层和计算器模式下，执行脚本中可访问的变量集）<sup>\[Python 文档——Python 教程\] » 6. 模块</sup>。

一个库可能由多个模块组成，不同的库可能使用相同名称的模块，Python 通过包来解决这个问题。

以下内容源自\[Python 文档——Python 教程\] » 6.4. 包

包是通过使用“带点号模块名”来构造 Python 模块命名空间的一种方式。 例如，模块名 A.B 表示名为 A 的包中名为 B 的子模块。 就像使用模块可以让不同模块的作者不必担心彼此的全局变量名一样，使用带点号模块名也可以让 NumPy 或 Pillow 等多模块包的作者也不必担心彼此的模块名冲突。

假设要为统一处理声音文件与声音数据设计一个模块集（“包”）。声音文件的格式很多 (通常以扩展名来识别，例如 .wav，.aiff，.au)，因此，为了不同文件格式之间的转换，需要创建和维护一个不断增长的模块集合。 为了实现对声音数据的不同处理（例如混声、添加回声、均衡器功能、创造人工立体声效果），还要编写无穷无尽的模块流。 你的包可能的架构是这样的（展示为多层级的文件系统形式）：
```
sound/                          最高层级的包
      __init__.py               初始化 sound 包
      formats/                  用于文件格式转换的子包
              __init__.py
              wavread.py
              wavwrite.py
              aiffread.py
              aiffwrite.py
              auread.py
              auwrite.py
              ...
      effects/                  用于音效的子包
              __init__.py
              echo.py
              surround.py
              reverse.py
              ...
      filters/                  用于过滤器的子包
              __init__.py
              equalizer.py
              vocoder.py
              karaoke.py
              ...
```      

需要有 `__init__.py` 文件才能让 Python 将包含该文件的目录当作包来处理（除非使用 namespace package，这是一个相对高级的特性）。 这可以防止重名的目录如 string 在无意中屏蔽后继出现在模块搜索路径中的有效模块。 在最简单的情况下，`__init__.py` 可以只是一个空文件，但它也可以执行包的初始化代码或设置 `__all__ `变量，这将在稍后详细描述<sup>\[Python 文档——Python 教程\] » 6.4. 包</sup>。

还可以从包中导入单个模块，例如：
```
import sound.effects.echo
```
这将加载子模块 sound.effects.echo。 它必须通过其全名来引用。
```
sound.effects.echo.echofilter(input, output, delay=0.7, atten=4)
```
另一种导入子模块的方法是 ：
```
from sound.effects import echo
```
这也会加载子模块 echo，并使其不必加包前缀，因此可按如下方式使用:
```
echo.echofilter(input, output, delay=0.7, atten=4)
```
Import 语句的另一种变体是直接导入所需的函数或变量：

```
from sound.effects.echo import echofilter
```
同样，这将加载子模块 echo，但这使其函数 echofilter() 直接可用:
```
echofilter(input, output, delay=0.7, atten=4)
```
注意，使用 from package import item 时，item 可以是包的子模块（或子包），也可以是包中定义的函数、类或变量等其他名称。import 语句首先测试包中是否定义了 item；如果未在包中定义，则假定 item 是模块，并尝试加载。如果找不到 item，则触发 ImportError 异常。

相反，使用 import item.subitem.subsubitem 句法时，除最后一项外，每个 item 都必须是包；最后一项可以是模块或包，但不能是上一项中定义的类、函数或变量。


#### 📚使用标准库示例1——查看当前用户

下面代码展示使用标注库的流程：
1. 导入需要的模块、函数、类或者变量；

````python
import os
def get_current_username():
    """跨平台获取当前用户名"""
    if os.name == 'nt':           # Windows
        return os.environ.get('USERNAME', '')
    else:                        # Linux / macOS
        return os.environ.get('LOGNAME') or os.environ.get('USER', '')

print(get_current_username())
````

#### 🧹Ruff Or Pylint

❗以下内容使用 Qwen3.5:9B 生成

Ruff 和 Pylint 都是 Python 生态中用于代码质量检查的重要工具，但两者在设计理念和底层实现上存在显著差异。最本质的区别在于它们的编写语言不同，Ruff 是用 Rust 编写的编译型程序，而 Pylint 则是用 Python 编写的解释型脚本。这一架构上的差异直接导致了两者在性能表现上的巨大鸿沟，Ruff 利用 Rust 的高性能特性，在处理大型代码库时的扫描速度通常比 Pylint 快几十倍甚至上百倍，能够在秒级内完成对整项目的分析，这对于追求开发效率的现代团队来说至关重要。

在功能覆盖方面，两者的定位也有所不同。Ruff 的设计初衷是成为一个轻量且全能的工具，它旨在替代包括 Flake8、Isort 和 Black 在内的多个传统工具，能够同时处理代码风格检查、导入排序以及格式化问题，从而减少项目中的依赖数量。相比之下，Pylint 作为一个老牌工具，更侧重于对代码逻辑的深入分析，例如计算圈复杂度或检测复杂的继承关系等深层问题，这些在 Ruff 当前版本中尚未完全覆盖的功能领域，Pylint 依然是行业内的权威参考。

从配置和集成角度来看，Ruff 采用了现代化的 TOML 格式配置文件，更加简洁直观且易于在 CI/CD 流程中自动化，支持一键修复大部分检查出的问题；而 Pylint 通常使用传统的 INI 或 Python 文件进行配置，虽然生态插件丰富，但设置过程相对繁琐。对于大多数开发者而言，如果项目更看重运行速度和构建体验，Ruff 无疑是首选方案，因为它能以最小的代价解决绝大多数常见的代码质量问题；若团队有特定的深度逻辑检查需求或必须兼容老旧的合规流程，则 Pylint 依然具备不可替代的价值。总体而言，当前趋势是倾向于用 Ruff 来统一 Python 项目的检查标准，仅在必要时保留部分 Pylint 规则以补充其分析深度的不足。


文件是一种在存储介质上持久化保存并组织数据的逻辑结构。然而，要让这一抽象概念转化为计算机可识别的物理信号，并确保数据在不同系统间具备可读性与兼容性，我们必须依赖特定的规则来定义其内部表示方式——这便是文件的编码和格式。

**从字符编码到图像像素，读懂计算机如何存储世界**

我们将通过一张分层的逻辑图（Mermaid）来直观展示文件从“抽象概念”到“物理存储”的转化过程

这张图表展示了数据如何在逻辑层、规则层和物理层之间转换，体现了“文件”的本质是信息的封装与标准化。

📖 详解：文件如何“翻译”世界

为了让你更好地理解这张图，我们可以将上述流程拆解为三个核心环节：

1. 🎯 逻辑抽象层 (Logical Layer)
*   概念：这是人类视角的“文件”。当你创建一个 `photo.jpg` 时，你脑海中有一个完整的图像。
*   作用：操作系统（如 Windows/Linux）将这一连串零散的数据块看作一个独立的实体，分配给它一个文件名和路径。此时它只是一个内存中的引用或目录表里的条目，尚未与物理硬件强绑定。

2. ⚙️ 规则层 (Rules Layer) —— 核心灵魂
这是连接抽象与现实的桥梁，也是你提到的“必须依赖的规则”。它包含两个关键部分：
*   文件格式 (Format)：
    *   定义：就像一封信的“信封格式”和“内部排版标准”。
    *   作用：告诉计算机“哪里是文件头（开始）”，“哪里是图片数据主体”，“哪里是结束标记”。
    *   例子：`JPEG` 规定了颜色压缩算法；`PDF` 规定了字体嵌入方式。没有格式，一串乱码无法被软件识别为特定文档。
*   编码 (Encoding)：
    *   定义：字符或物理量与数字的映射规则。
    *   作用：将人类的“符号”（如汉字 '中'）或“颜色值”（如红色）翻译成计算机唯一的“语言”（二进制）。
    *   例子：`UTF-8` 把 '中' 变成 `E4 B8 AD`；`RGB` 把红色像素 (255,0,0) 变成 `11111111 00000000 ...`。

3. 💾 物理介质层 (Physical Layer)
*   概念：这是现实世界的“文件”。
*   作用：所有的规则最终都坍缩为对物理信号的操纵。
    *   硬盘：通过改变磁盘表面的磁性极性（0 vs 1）。
    *   SSD/Flash：控制存储单元中的电荷有无。
    *   云存储：将数据打包成二进制包，通过网络协议传输并落地到服务器物理盘上。


💡 为什么需要“编码”与“格式”？（兼容性的本质）

图表中最关键的一环是 “确保不同系统间的可读性”。如果没有规则和编码，计算机世界将是孤岛：

1.  通用语言：
    *   `UTF-8` 编码使得一个包含中、英、法文字符的文件，既可以在 Windows 上打开，也能在 Linux 终端解析。它消除了“字符集不匹配”导致的乱码（Mojibake）。
2.  标准化结构：
    *   `MP4` 格式被所有播放器识别。即便是一个手机 App 生成的视频，只要遵循 MP4 的容器规范（Header, Index, Stream），任何符合标准的解码器都能还原它。
3.  容错与恢复：
    *   物理存储会损坏（坏道、掉电）。编码规则中的校验位（如 CRC）允许系统在读取时自我修复，确保“逻辑结构”的完整性不被物理噪声破坏。

🚀 总结
文件不仅仅是数据，它是人类意图被标准化后的二进制映射。
*   格式 (Format) 是骨架，决定了数据的组织结构；
*   编码 (Encoding) 是血液，定义了信息的数值表达；
*   存储介质 (Storage) 是土壤，承载了最终的物理信号。

只有当这三者完美协同时，存储在硬盘深处那串无序的 `0` 和 `1`，才能在屏幕上重新变成你熟悉的字符、照片和视频。

关于文本文件的换行符，通常有两种标准：

1.  `\r\n` (CRLF)：这是 Windows 系统传统的标准。
2.  `\n` (LF)：这是 Unix/Linux/macOS 系统的标准，也是目前互联网开发（如 Web、跨平台工具）的通用标准。

💡 建议与说明

为了最佳兼容性和现代开发习惯，通常推荐统一使用 `\n`。原因如下：
*   跨平台兼容性更好：大多数编辑器（VS Code, Notepad++, Vim 等）和编程语言都能自动处理或忽略 `\r\n` 中的 `\r`。
*   代码简洁：在生成代码、数据文件时，使用 `\n` 更节省字节空间。
*   系统转换：现代 Windows 系统和编辑器通常会自动将接收到的 `\n` 转换为本地格式（`\r\n`），而反之则不一定（CRLF 有时会被误判为其他控制字符）。

### ✍️伴读札记

#### 📂优先使用 pathlib

`pathlib` 在 Python 3.4+ 中已引入，是现代文件系统路径操作的首选方式，但它不能完全替代 `os` 和 `os.path`。两者功能定位不同：`pathlib` 专注于路径本身的结构化操作，而 `os` 模块提供系统级、底层接口。以下是对比总结及最佳实践建议：

✅ pathlib 的优势（推荐用于路径处理）

- 面向对象设计：使用 `Path` 类，语法更清晰直观
- 跨平台兼容：自动适配 Windows/Linux/macOS 的路径分隔符和大小写处理
- 链式调用支持：支持多步操作如 `.parent / .name / suffix`
- 现代 Python 推荐方式**：Python 官方文档建议使用 `pathlib`

示例：
```python
from pathlib import Path
p = Path("data") / "files" / "input.txt"
if p.exists() and not p.is_dir():
    print(f"{p} is a file")
```
⚠️ os 模块无法完全被替代的场景

`os` 提供的是更底层的系统级操作，而 `pathlib` 不覆盖这些功能：

| 功能类别           | 示例代码                      | pathlib 是否支持 |
|--------------------|-------------------------------|------------------|
| 环境变量访问       | `os.environ['PATH']`          | ❌               |
| 系统命令执行       | `os.system("ls -la")`         | ❌               |
| 文件权限/所有权操作 | `os.chmod()`, `os.chown()`    | ❌（需调用 os）   |
| 状态信息获取       | `os.stat()`, `os.lstat()`     | ✅（有 Path.stat）|
| 文件锁机制         | `fcntl.flock()`, `msvcrt`     | ❌               |
| 符号链接相关操作   | `os.symlink()`, `os.readlink()`| ⚠️部分支持       |

示例：
```python
import os
from pathlib import Path

# pathlib 无法直接替换这些系统调用
os.chmod("file.txt", 0o644)           # 修改权限
os.environ["PATH"] = "/usr/bin"        # 设置环境变量
os.system("echo 'hello'")              # 执行系统命令
```
🧭 总结建议

- ✅ 优先使用 `pathlib`：处理路径构建、遍历、读取等操作时首选。
- ⚠️ 保留 `os` 模块功能：权限管理、环境变量、系统调用等仍需依赖 `os`。
- 📦 可混合使用：将 `Path` 与 `os` API 结合，实现更灵活的文件系统操作。
- 🔐 Python 2/3兼容性：如果仍在 Python 2 中运行项目，无法使用 `pathlib`。

## 习题
### 一、选择题
1\. 打开文件时分别采用以下读写模式，只能进行读不能写的是（）。

A. w

B. a

C. w+

D. r

🏍️解答：D

- `r`：只读，只可读不可写，文件不存在时报错
- `w`：只写，打开会清空文件原有内容
- `a`：追加，在文件末尾写入数据
- `w+`：读写，可读可写，打开清空原文件

2\. 执行语句 fo = open(Sample.txt, 'r+') 后，对文件 Sample.txt 能够进行的操作是（）。

A. 只能写不能读

B. 只能读不能写

C. 既可以读，也可以写

D. 不能读，也不能写

🏍️解答：C

在 Python 中，`'r+'` 模式表示以**读写**方式打开文件。文件指针会放在文件的开头，并且要求文件必须存在（如果文件不存在，会抛出 `FileNotFoundError` 错误）。

与其他模式对比：
*   A (w)：只写，文件不存在则创建，存在则覆盖。
*   B (a)：只写（追加），文件不存在则创建。
*   D (r)：只读，文件必须存在。
*   C (r+)：读写，文件必须存在。

1\. 什么是文本文件？文本文件和二进制文件各自的存储特点是什么？

文本文件是一种以字符编码（如 ASCII、UTF-8、GBK 等）形式存储数据的文件。它由一系列字符（包括字母、数字、标点符号和换行符）组成，其内容本质上是对人类可读字符的二进制编码。

文本文件：
存储的是字符的编码值（如ASCII或UTF-8），读写时需要进行编码/解码转换；内容人类可直接阅读；在不同操作系统下，换行符（\n与\r\n）可能会被自动转换。

二进制文件：
存储的是数据在内存中的原始二进制位（原样存储），读写时无需编码转换；内容人类无法直接读懂（通常显示为乱码）；不处理换行符，所有字节均按原样原序存储，读写效率更高且不会丢失精度。

2\. 请说明 write() 和 writelines() 方法的区别。

针对 `write()` 和 `writelines()` 的区别，简述如下：

- 参数不同：`write()` 写入单个字符串；`writelines()` 写入字符串列表（或任意可迭代对象）。
- 本质相同：`writelines()` 内部本质是循环调用 `write()`，用于批量写入。
- 核心注意点（共性）：两者都不会自动换行。若需换行，都必须手动在字符串末尾添加 `\n`。

4\. 文本文件和二进制文件的读写操作有何不同？

文本文件与二进制文件读写操作的核心区别可简述如下：

1. 数据类型不同：文本模式操作字符串（`str`），二进制模式操作字节串（`bytes`）。
2. 编码转换不同：文本模式自动进行编码解码（需指定 `encoding`，如 `utf-8`），二进制模式完全不转换，原样保留原始字节。
3. 换行符处理不同：文本模式会自动将换行符转换为系统标准（如 Windows 的 `\r\n` 与 Linux 的 `\n` 互转）；二进制模式不做任何转换，保持原样。
4. 适用场景不同：文本模式用于 `.txt`、`.json` 等人可读文件；二进制模式用于图片、音视频、压缩包等非文本文件（若不使用二进制模式，数据会被损坏）。

一句话概括：文本读写“懂”编码和换行，二进制读写“只看”原始字节。

3\. 从第 2 题的 file1.txt 文件中读出数据存入列表 a 中，从中挑出所有个位和百位是偶数的数据存入 b 列表中，并存放进 CSV 文件 file2.csv 中。

````python
import csv

# 1. 从 file1.txt 读取数据存入列表 a
a = []
with open('data/file1.txt', 'r', encoding='utf-8') as f:
    for line in f:
        num = int(line.strip())   # 去除换行符并转为整数
        a.append(num)

# 2. 筛选：个位和百位都是偶数
b = []
for num in a:
    units = num % 10           # 个位
    hundreds = (num // 100) % 10  # 百位
    if units % 2 == 0 and hundreds % 2 == 0:
        b.append(num)

# 3. 将 b 写入 CSV 文件 file2.csv（每行一个数字）
with open('data/file2.csv', 'w', newline='', encoding='utf-8') as f:
    writer = csv.writer(f)
    for num in b:
        writer.writerow([num])   # 写入一行，单列

print(f"筛选出 {len(b)} 个符合条件的数字，已写入 file2.csv")
````

4\. 准备一篇英文文章 file3.txt（至少 3 段），编程读出其中的内容，将所有字符进行替代，替代关系为 f(p) = (p * 11) mod 256。p 是某个字符的 ASCII 值，f(p)是计算后新字符的 ASCII 值。如果计算后 f(p)的值小于等于 32 或大于 130，则该字符不变，否则用 f(p)所对应的字符替换原有字符后写入文件 file4.txt 中。

````python
# 读取 file3.txt 内容
with open('data/file3.txt', 'r', encoding='utf-8') as f:
    content = f.read()

# 定义转换函数
def transform_char(ch):
    p = ord(ch)                 # 获取字符的 ASCII 码
    f = (p * 11) % 256          # 计算新 ASCII 码
    if f <= 32 or f > 130:
        return ch               # 条件不满足，保留原字符
    else:
        return chr(f)           # 替换为新字符

# 对每个字符进行转换
new_content = ''.join(transform_char(c) for c in content)

# 写入 file4.txt
with open('data/file4.txt', 'w', encoding='utf-8') as f:
    f.write(new_content)

print("文件已处理并保存为 file4.txt")
````

5\. 有一个 20 名学生的成绩单，包含每个学生的姓名和 5 门课程成绩。姓名依次记作 A～T，成绩均随机产生（范围为 50～100 分），请计算出平均成绩，并按平均成绩由高到低排序后分别写入文本文件 file5.txt 和 CSV 文件 file6.csv 中。

````python
import random
import csv

# 1. 生成20名学生成绩（姓名 A~T）
students = []
names = [chr(ord('A') + i) for i in range(20)]   # ['A', 'B', ..., 'T']

for name in names:
    scores = [random.randint(50, 100) for _ in range(5)]   # 5门课程成绩
    avg = sum(scores) / 5
    students.append({
        'name': name,
        'scores': scores,
        'average': avg
    })

# 2. 按平均成绩由高到低排序
students_sorted = sorted(students, key=lambda x: x['average'], reverse=True)

# 3. 写入文本文件 file5.txt
with open('data/file5.txt', 'w', encoding='utf-8') as f:
    f.write("姓名\t课程1\t课程2\t课程3\t课程4\t课程5\t平均分\n")
    for stu in students_sorted:
        scores_str = '\t'.join(map(str, stu['scores']))
        f.write(f"{stu['name']}\t{scores_str}\t{stu['average']:.2f}\n")

# 4. 写入 CSV 文件 file6.csv
with open('data/file6.csv', 'w', newline='', encoding='utf-8') as f:
    writer = csv.writer(f)
    # 写入表头
    writer.writerow(['姓名', '课程1', '课程2', '课程3', '课程4', '课程5', '平均分'])
    for stu in students_sorted:
        row = [stu['name']] + stu['scores'] + [round(stu['average'], 2)]
        writer.writerow(row)

print("成绩已生成并排序，分别写入 file5.txt 和 file6.csv")
````

8\. 编程将李成每月工资记账信息（年月信息、收入、支出）写入文件 expence.csv 中，并计算每年的余额信息。

````python
import csv
import random
from datetime import datetime, timedelta

# ---------- 1. 生成模拟数据（2020-01 至 2022-12） ----------
start_date = datetime(2020, 1, 1)
end_date = datetime(2022, 12, 1)
data = []  # 存放 (年月, 收入, 支出)

current = start_date
while current <= end_date:
    year_month = current.strftime("%Y-%m")
    income = random.randint(5000, 10000)       # 月收入 5000~10000
    expense = random.randint(2000, 6000)       # 月支出 2000~6000
    data.append((year_month, income, expense))
    # 下个月
    if current.month == 12:
        current = datetime(current.year + 1, 1, 1)
    else:
        current = datetime(current.year, current.month + 1, 1)

# ---------- 2. 计算每月余额，写入 expence.csv ----------
with open('data/expence.csv', 'w', newline='', encoding='utf-8') as f:
    writer = csv.writer(f)
    writer.writerow(['年月', '收入', '支出', '月余额'])   # 表头
    for ym, inc, exp in data:
        balance = inc - exp
        writer.writerow([ym, inc, exp, balance])

print("每月记账明细已写入 expence.csv")

# ---------- 3. 计算每年余额信息（按年汇总） ----------
yearly_summary = {}   # key: 年份, value: [年总收入, 年总支出, 年余额]
for ym, inc, exp in data:
    year = ym.split('-')[0]
    if year not in yearly_summary:
        yearly_summary[year] = [0, 0, 0]
    yearly_summary[year][0] += inc
    yearly_summary[year][1] += exp
    yearly_summary[year][2] += (inc - exp)

# 打印年度汇总
print("\n=== 每年余额汇总 ===")
for year in sorted(yearly_summary.keys()):
    total_inc, total_exp, total_bal = yearly_summary[year]
    print(f"{year}年：总收入 {total_inc}，总支出 {total_exp}，年余额 {total_bal}")
````

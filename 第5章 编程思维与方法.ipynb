符号计算库（如 SymPy）延续纸笔的解析思维，能够完成符号层面的微分、积分、方程求解、公式化简、矩阵运算，全程保留抽象的数学符号，输出解析表达式。它等价于在计算机上做严谨的理论推导，可直接复现课本上的公式推演，用来核对手工推导结果、推导复杂物理模型的解析解，弥补纸笔推导中容易算错、演算繁琐的短板。但面对高阶非线性、多耦合物理系统时，多数问题不存在简洁解析解，符号计算便会遇到瓶颈。

1\. 求自然对数 $e$ 的近似值，当任意项的值小于 $10^{-4}$ 时结束计算，近似公式为：
$$
e \approx 1+\frac{1}{1!}+\frac{1}{2!}+\frac{1}{3!}+\dots+\frac{1}{n!}
$$

```python
tol = 1e-4          # 精度阈值
e_approx = 1.0      # 初始值，对应 1/0!
term = 1.0          # 当前项，初始为 1/0!
k = 1               # 阶乘分母的基数

while True:
    term /= k       # 计算 1/k!（k 从 1 开始）
    if term < tol:  # 如果当前项小于阈值，停止累加
        break
    e_approx += term
    k += 1

print(f"e 的近似值: {e_approx}")
```

2\. 计算机定积分：求函数 $ \int_{0}^{1} e^{-x^2} \mathrm{d}x$ 的近似值。

```python
import math

# 被积函数
def f(x):
    return math.exp(-x * x)

# 积分区间
a, b = 0.0, 1.0
n = 1000          # 等分数（必须为偶数）
h = (b - a) / n

# 辛普森公式：h/3 * [f(a) + f(b) + 4*sum(奇数点) + 2*sum(偶数点)]
s = f(a) + f(b)
for i in range(1, n):
    x = a + i * h
    if i % 2 == 0:
        s += 2 * f(x)
    else:
        s += 4 * f(x)

integral = s * h / 3
print(f"定积分近似值: {integral}")
```

```python
x = sp.symbols('x')          # 定义符号变量
f = sp.exp(-x**2)            # 被积函数

# 正确格式：(x, 0, 1) 而不是 (0.999, 0, 1) 或其他
integral_expr = sp.integrate(f, (x, 0, 1))

print("符号结果:", integral_expr)
print("数值结果:", integral_expr.evalf())
```

3\. 编程求1000到1100的所有质数，每行输出6个，分多行输出。

```python
def is_prime(n):
    """判断 n 是否为质数"""
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

# 收集质数
primes = [n for n in range(1000, 1101) if is_prime(n)]

# 每行输出6个，用空格分隔
for i in range(0, len(primes), 6):
    row = primes[i:i+6]
    print(' '.join(map(str, row)))
```

4\. 一个自然数倒过来读仍是这个数，就叫回文数，如151。编程求出100~999的回文数。

```python
# 方法：遍历所有三位数，判断字符串反转是否相同
palindromes = []
for n in range(100, 1000):
    if str(n) == str(n)[::-1]:
        palindromes.append(n)

# 输出（每行10个，便于查看）
for i in range(0, len(palindromes), 10):
    print(' '.join(map(str, palindromes[i:i+10])))
```

5\. “四叶玫瑰数”是指一个4位数，其各位数字的4次方之和等于该数本身。例如，1634（$1^4+6^4+3^4+4^4 = 1634$）。编程求解所有的四叶玫瑰数。

```python
# 遍历所有四位数
rose_numbers = []
for num in range(1000, 10000):
    # 拆分为各位数字
    a = num // 1000          # 千位
    b = (num // 100) % 10    # 百位
    c = (num // 10) % 10     # 十位
    d = num % 10             # 个位
    if a**4 + b**4 + c**4 + d**4 == num:
        rose_numbers.append(num)

# 输出结果
print("四叶玫瑰数有：", rose_numbers)
```

6\. 马克思手稿中有一道趣味数学问题：有30个人，其中有男人、女人和小孩，在一家饭馆吃饭花了50先令；每个男人花3先令，每个女人花2先令，每个小孩花1先令；男人、女人和小孩各有几人？

```python
# 遍历所有可能的男人人数 x 和女人人数 y
solutions = []
for x in range(31):               # 男人最多30人
    for y in range(31 - x):       # 女人最多剩余人数
        z = 30 - x - y
        if 3*x + 2*y + z == 50:
            solutions.append((x, y, z))

print("所有可能的组合（男人, 女人, 小孩）：")
for sol in solutions:
    print(f"男人: {sol[0]:2d}人, 女人: {sol[1]:2d}人, 小孩: {sol[2]:2d}人")
```

7\. 设有一个分数序列：$\frac{2}{1},\frac{3}{2},\frac{5}{3},\frac{8}{5},\frac{13}{8},\frac{21}{13},\dots$，编程求这个数列前20项之和。

```python
# 初始分子分母
a, b = 2, 1
total = 0.0

for _ in range(20):
    total += a / b          # 累加当前项
    a, b = a + b, a         # 更新为下一项：新分子=原分子+原分母，新分母=原分子

print("前20项之和:", total)
```

8\. 用牛顿迭代法求方程 $2x^3-4x^2+3x-6=0$ 在1.5附近的根。

```python
def f(x):
    return 2*x**3 - 4*x**2 + 3*x - 6

def df(x):
    return 6*x**2 - 8*x + 3

# 初始猜测
x0 = 1.5
tol = 1e-10
max_iter = 100

for i in range(max_iter):
    fx = f(x0)
    dfx = df(x0)
    if abs(fx) < tol:
        break
    x1 = x0 - fx / dfx
    if abs(x1 - x0) < tol:
        x0 = x1
        break
    x0 = x1

print(f"方程的根近似为: {x0:.10f}")
```

9\. 用迭代法求 $x=\sqrt{a}$，求平方根的迭代公式为 $x_{n+1}=\frac12\left(x_n+\frac{a}{x_n}\right)$。

```python
a = 5.0
x = a / 2.0
tol = 1e-12
max_iter = 100

for i in range(max_iter):
    x_new = 0.5 * (x + a / x)
    if abs(x_new - x) < tol:
        x = x_new
        break
    x = x_new

print(f"√{a} ≈ {x:.12f}")
```

```python
# 模拟输入：某班学生成绩（人数和具体分数）
scores = [85, 92, 78, 90, 88, 76, 95, 89, 84, 91]  # 10个学生

# 按从高到低排序
scores_sorted = sorted(scores, reverse=True)

print("按从高到低排序的成绩：")
for i, score in enumerate(scores_sorted, 1):
    print(f"第{i:2d}名: {score}分")
```

```python
# 模拟输入：某班学生成绩（人数和分数）
scores = [85, 92, 78, 90, 88, 76, 95, 89, 84, 91]  # 10个学生
n = len(scores)

# 冒泡排序（从高到低，降序）
for i in range(n - 1):
    for j in range(0, n - i - 1):
        if scores[j] < scores[j + 1]:   # 若前小后大，交换（降序）
            scores[j], scores[j + 1] = scores[j + 1], scores[j]

# 输出结果
print("按从高到低排序的成绩：")
for i, score in enumerate(scores, 1):
    print(f"第{i:2d}名: {score}分")
```

13\. 求解数学灯谜。设有以下算式：
$$
\begin{array}{r}
\ \ A\ B\ C\ D\\
-\ \ C\ D\ C\\
\hline
\ \ A\ B\ C
\end{array}
$$
编程求 $A、B、C、D$ 的值。



```python
# A（1-9），B（0-9），C（1-9，不能为0），D（0-9）
solutions = []
for A in range(1, 10):
    for B in range(10):
        for C in range(1, 10):   # C 从1开始
            for D in range(10):
                abcd = 1000*A + 100*B + 10*C + D
                cdc  = 100*C + 10*D + C
                abc  = 100*A + 10*B + C
                if abcd - cdc == abc:
                    solutions.append((A, B, C, D))

print("所有解（A, B, C, D）：")
for sol in solutions:
    print(f"A={sol[0]}, B={sol[1]}, C={sol[2]}, D={sol[3]}")
```

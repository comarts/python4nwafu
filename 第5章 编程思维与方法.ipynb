**超越纸笔局限：以 Python 拓展数学与物理的探究边界**

符号计算库（如 SymPy）延续纸笔的解析思维，能够完成符号层面的微分、积分、方程求解、公式化简、矩阵运算，全程保留抽象的数学符号，输出解析表达式。它等价于在计算机上做严谨的理论推导，可直接复现课本上的公式推演，用来核对手工推导结果、推导复杂物理模型的解析解，弥补纸笔推导中容易算错、演算繁琐的短板。但面对高阶非线性、多耦合物理系统时，多数问题不存在简洁解析解，符号计算便会遇到瓶颈。

数值计算库（如 NumPy、SciPy）则走离散求解的路径，将数学模型转化为数值运算，处理符号方法难以解决的实际问题：数值积分、常微分 / 偏微分方程仿真、矩阵数值求解、数据拟合、物理过程模拟。它不追求完美的解析公式，而是给出具体条件下的数值结果，能够处理现实物理中带参数、带边界条件的复杂模型。

```python
f = x**3 * sp.exp(x) # f(x) = x^3 * e^x
df = sp.diff(f, x)   # 自动求导，无需手动使用乘积法则
f, df
```
